```python
from optbinning import OptimalBinning
```

#### 1. count_issued

```python
optb_count_issue = OptimalBinning(name='count_issued', dtype="numerical", solver="cp")
```

```python
data = df_retention_trave.copy()
x = data['count_issued'].values
y = data['is_churn']
optb_count_issue.fit(x, y)
```

```python
result = optb_count_issue.splits
print("Optimal Splits : ", result)
```

```python
optb_count_issue.binning_table.build() 
```

```python
optb_count_issue.binning_table.plot(metric="woe")
```
