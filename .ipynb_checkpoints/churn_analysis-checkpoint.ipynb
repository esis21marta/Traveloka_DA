```python
def plot_cm(ax, y_true, y_pred, classes, title, th=0.5, cmap=plt.cm.Blues):
    
    cm = confusion_matrix(y_true, y_pred)
    
    im = ax.imshow(cm, interpolation='nearest', cmap=cmap)
    ax.set_title(title)

    tick_marks = np.arange(len(classes))
    ax.set_xticks(tick_marks)
    ax.set_yticks(tick_marks)
    ax.set_xticklabels(classes)
    ax.set_yticklabels(classes)

    thresh = cm.max() / 2.
    for i, j in itertools.product(range(cm.shape[0]), range(cm.shape[1])):
        ax.text(j, i, cm[i, j],
                 horizontalalignment="center",
                 color="white" if cm[i, j] > thresh else "black")
    ax.set_ylabel('Actual')
    ax.set_xlabel('Predicted')
```

# Create churn target

People who churn are people who haven't bought anything in last one year (12 Oct 2018 - 12 Oct 2019)

```python
# create customer churn using last booking time
df_churn_all = pd.DataFrame(pd.DataFrame(customer_data.groupby('profile_id')['last_booking_time'].max())['last_booking_time']\
                .apply(lambda time: 1 if time<pd.to_datetime('2018-10-12' )else 0))\
                .reset_index()\
                .rename(columns={'last_booking_time':'churn'})
```

```python
df_churn_all.head()
```

# Feature engineering

### create the features

```python
customer_data.dtypes
```

```python
# add first issued - first book time, to see the flow, did she just book and not proceed to the next step of payment?
customer_data['delta_first_time'] = ((customer_data['first_issued_time']-customer_data['first_booking_time'])/np.timedelta64(1,'h'))
customer_data.loc[customer_data['delta_first_time']<0,'delta_first_time'] = 0
customer_data['delta_first_time'] = customer_data['delta_first_time'].fillna(customer_data['delta_first_time'].median())

customer_data['last_issued_days'] = (pd.to_datetime('2018-10-12')-customer_data['last_issued_time']).dt.days
customer_data['last_issued_days'] = customer_data['last_issued_days'].fillna(customer_data['last_issued_days'].median())
```

```python
# create customer churn for each category using last booking time
customer_data['churn'] = customer_data['last_booking_time'].apply(lambda time: 1 if time<pd.to_datetime('2018-10-12' )else 0)
```

```python
customer_data.head(10)
```

```python
# numerical data binned, turned into categorical with one hot encoder
# numerical with nan
df_last_issued_days = customer_data.groupby(['profile_id','primary_sales_product_type'])['last_issued_days'].sum()\
                        .unstack()\
                        .reset_index()\
                        .rename(columns={'FLIGHT':'last_issued_days_flight'\
                                         ,'HOTEL':'last_issued_days_hotel'\
                                         ,'TRAIN':'last_issued_days_train'\
                                         ,'TRIP':'last_issued_days_trip'})

df_last_issued_days['last_issued_days_flight'] = pd.cut(df_last_issued_days['last_issued_days_flight'], [0,180,365,730,1095,np.inf], include_lowest=True).cat.codes
df_last_issued_days['last_issued_days_hotel'] = pd.cut(df_last_issued_days['last_issued_days_hotel'], [0,180,365,730,1095,np.inf], include_lowest=True).cat.codes
df_last_issued_days['last_issued_days_train'] = pd.cut(df_last_issued_days['last_issued_days_train'], [0,180,365,730,1095,np.inf], include_lowest=True).cat.codes
df_last_issued_days['last_issued_days_trip'] = pd.cut(df_last_issued_days['last_issued_days_trip'], [0,180,365,730,1095,np.inf], include_lowest=True).cat.codes

df_last_issued_days.loc[df_last_issued_days['last_issued_days_flight'] < 0, 'last_issued_days_flight'] = 99
df_last_issued_days.loc[df_last_issued_days['last_issued_days_hotel'] < 0, 'last_issued_days_hotel'] = 99
df_last_issued_days.loc[df_last_issued_days['last_issued_days_train'] < 0, 'last_issued_days_train'] = 99
df_last_issued_days.loc[df_last_issued_days['last_issued_days_trip'] < 0, 'last_issued_days_trip'] = 99


df_delta_first_time = customer_data.groupby(['profile_id','primary_sales_product_type'])['delta_first_time'].sum()\
                        .unstack()\
                        .reset_index()\
                        .rename(columns={'FLIGHT':'delta_first_time_flight'\
                                         ,'HOTEL':'delta_first_time_hotel'\
                                         ,'TRAIN':'delta_first_time_train'\
                                         ,'TRIP':'delta_first_time_trip'})

df_delta_first_time['delta_first_time_flight'] = pd.cut(df_delta_first_time['delta_first_time_flight'], [0,5,12,24,48,np.inf], include_lowest=True).cat.codes
df_delta_first_time['delta_first_time_hotel'] = pd.cut(df_delta_first_time['delta_first_time_hotel'], [0,5,12,24,48,np.inf], include_lowest=True).cat.codes
df_delta_first_time['delta_first_time_train'] = pd.cut(df_delta_first_time['delta_first_time_train'], [0,5,12,24,48,np.inf], include_lowest=True).cat.codes
df_delta_first_time['delta_first_time_trip'] = pd.cut(df_delta_first_time['delta_first_time_trip'], [0,5,12,24,48,np.inf], include_lowest=True).cat.codes

df_delta_first_time.loc[df_delta_first_time['delta_first_time_flight'] < 0, 'delta_first_time_flight'] = 99
df_delta_first_time.loc[df_delta_first_time['delta_first_time_hotel'] < 0, 'delta_first_time_hotel'] = 99
df_delta_first_time.loc[df_delta_first_time['delta_first_time_train'] < 0, 'delta_first_time_train'] = 99
df_delta_first_time.loc[df_delta_first_time['delta_first_time_trip'] < 0, 'delta_first_time_trip'] = 99
```

```python
# one hot encoding
k = 0 
df_delta_first_time_1 = pd.DataFrame()
for col in df_delta_first_time.drop(columns=['profile_id']).columns.values:
    if k == 0:
        df_delta_first_time_1 = pd.get_dummies(df_delta_first_time.set_index('profile_id')[col],prefix=col+'_')
    elif k>0:
        df_delta_first_time_1 = pd.concat([df_delta_first_time_1,pd.get_dummies(df_delta_first_time.set_index('profile_id')[col],prefix=col+'_')],axis=1)
    k=k+1
    
df_delta_first_time_1 = df_delta_first_time_1.reset_index()
```

```python
# one hot encoding
k = 0 
df_last_issued_days_1 = pd.DataFrame()
for col in df_last_issued_days.drop(columns=['profile_id']).columns.values:
    if k == 0:
        df_last_issued_days_1 = pd.get_dummies(df_last_issued_days.set_index('profile_id')[col],prefix=col+'_')
    elif k>0:
        df_last_issued_days_1 = pd.concat([df_last_issued_days_1,pd.get_dummies(df_last_issued_days.set_index('profile_id')[col],prefix=col+'_')],axis=1)
    k=k+1

df_last_issued_days_1 = df_last_issued_days_1.reset_index()
```

```python
df_churn = customer_data.groupby(['profile_id','primary_sales_product_type'])['churn'].sum()\
                        .unstack()\
                        .fillna(2)\
                        .reset_index()\
                        .rename(columns={'FLIGHT':'churn_booking_flight'\
                                         ,'HOTEL':'churn_booking_hotel'\
                                         ,'TRAIN':'churn_booking_train'\
                                         ,'TRIP':'churn_booking_trip'})
```

```python
# numerical features
df_count_booking = customer_data.groupby(['profile_id','primary_sales_product_type'])['count_booking'].sum()\
                        .unstack()\
                        .fillna(0)\
                        .reset_index()\
                        .rename(columns={'FLIGHT':'count_booking_flight'\
                                         ,'HOTEL':'count_booking_hotel'\
                                         ,'TRAIN':'count_booking_train'\
                                         ,'TRIP':'count_booking_trip'})

df_count_issued = customer_data.groupby(['profile_id','primary_sales_product_type'])['count_issued'].sum()\
                        .unstack()\
                        .fillna(0)\
                        .reset_index()\
                        .rename(columns={'FLIGHT':'count_issued_flight'\
                                         ,'HOTEL':'count_issued_hotel'\
                                         ,'TRAIN':'count_issued_train'\
                                         ,'TRIP':'count_issued_trip'})

df_avg_issued = customer_data.groupby(['profile_id','primary_sales_product_type'])['avg_issued_invoice_amount_idr'].sum()\
                        .unstack()\
                        .fillna(0)\
                        .reset_index()\
                        .rename(columns={'FLIGHT':'avg_issued_flight'\
                                         ,'HOTEL':'avg_issued_hotel'\
                                         ,'TRAIN':'avg_issued_train'\
                                         ,'TRIP':'avg_issued_trip'})

df_avg_booking = customer_data.groupby(['profile_id','primary_sales_product_type'])['avg_booking_invoice_amount_idr'].sum()\
                        .unstack()\
                        .fillna(0)\
                        .reset_index()\
                        .rename(columns={'FLIGHT':'avg_booking_flight'\
                                         ,'HOTEL':'avg_booking_hotel'\
                                         ,'TRAIN':'avg_booking_train'\
                                         ,'TRIP':'avg_booking_trip'})
```

```python
df_all = df_avg_booking.set_index('profile_id').join(df_churn.set_index('profile_id'))\
                    .join(df_avg_issued.set_index('profile_id'))\
                    .join(df_count_issued.set_index('profile_id'))\
                    .join(df_count_booking.set_index('profile_id'))\
                    .join(df_delta_first_time_1.set_index('profile_id'))\
                    .join(df_last_issued_days_1.set_index('profile_id'))\
                    .join(df_churn_all.set_index('profile_id'))
```

```python
df_all.columns
```

# Churn analysis with XGBoost

```python
df_all["churn"] = df_all["churn"].astype(int)
y = df_all["churn"].values
X = df_all.drop(columns=['churn'])


X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=101)
```

```python
model = XGBClassifier()
model.fit(X_train,y_train)
```

```python
# make predictions for test data

y_pred = model.predict(X_test)
threshold = 0.25

# predictions = [round(value) for value in y_pred]
predictions = []

for value in y_pred:
    if value<threshold:
        predictions.append(0)
    else :
        predictions.append(1)
# evaluate predictions
accuracy = accuracy_score(y_test, predictions)
print("Accuracy: %.2f%%" % (accuracy * 100.0))
```

```python
fpr, tpr, thresholds = metrics.roc_curve(y_test, predictions)
metrics.auc(fpr, tpr)
```

```python
# plot feature importance
plot_importance(model)
pyplot.show()
```

```python
y_pred
```

```python
threshold = 0.25

fig,ax = plt.subplots(1,2)
fig.set_size_inches(15,5)
plot_cm(ax[0], y_test, predictions, [0,1], 'Confusion Matrix', th=0.5, cmap=plt.cm.Blues)
```
